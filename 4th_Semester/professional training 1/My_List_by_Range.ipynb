```python
# Building data
group_name = ('Group A', 'Group B', 'Group C')
student_name_A = ('Mike', 'Jack', 'Emma', 'Lucy', 'Noah')
student_name_B = ('Oliver', 'Liam', 'Ava', 'Sophia', 'Mia')
student_name_C = ('Charlie', 'Ethan', 'Olivia', 'Lily', 'James')
score1 = [79, 68, 77, 86, 92]
score2 = [81, 83, 62, 76, 82]
score3 = [96, 73, 76, 81, 82]

# Create list of student names and scores for each group
groups_students = [student_name_A, student_name_B, student_name_C]
groups_scores = [score1, score2, score3]

# Print score report for each group
for i in range(len(group_name)):
    print("=" * 50)
    print(f"{group_name[i]} - Score Report")
    print("=" * 50)
    print(f"{'Student Name':<15} {'Score':<10}")
    print("-" * 50)
    
    # Print each student's score
    for j in range(len(groups_students[i])):
        print(f"{groups_students[i][j]:<15} {groups_scores[i][j]:<10}")
    
    # Calculate and print statistics
    total = sum(groups_scores[i])
    average = total / len(groups_scores[i])
    highest = max(groups_scores[i])
    lowest = min(groups_scores[i])
    
    print("-" * 50)
    print(f"{'Total Score:':<15} {total:<10}")
    print(f"{'Average Score:':<15} {average:.2f}")
    print(f"{'Highest Score:':<15} {highest:<10}")
    print(f"{'Lowest Score:':<15} {lowest:<10}")
    print()
```

```python
# Building data
group_name = ('Group A', 'Group B', 'Group C')
student_name_A = ('Mike', 'Jack', 'Emma', 'Lucy', 'Noah')
student_name_B = ('Oliver', 'Liam', 'Ava', 'Sophia', 'Mia')
student_name_C = ('Charlie', 'Ethan', 'Olivia', 'Lily', 'James')
score1 = [79, 68, 77, 86, 92]
score2 = [81, 83, 62, 76, 82]
score3 = [96, 73, 76, 81, 82]

# Combine all students and scores into a single list
all_students = []
all_scores = []

# Add Group A
for i in range(len(student_name_A)):
    all_students.append(student_name_A[i])
    all_scores.append(score1[i])

# Add Group B
for i in range(len(student_name_B)):
    all_students.append(student_name_B[i])
    all_scores.append(score2[i])

# Add Group C
for i in range(len(student_name_C)):
    all_students.append(student_name_C[i])
    all_scores.append(score3[i])

# Create list of tuples (student, score) and sort by score in descending order
student_score_pairs = list(zip(all_students, all_scores))
student_score_pairs.sort(key=lambda x: x[1], reverse=True)

# Calculate total score of the entire class
total_class_score = sum(all_scores)

# Print class summary
print("=" * 50)
print("CLASS SCORE REPORT - DESCENDING ORDER")
print("=" * 50)
print(f"{'Rank':<6} {'Student Name':<15} {'Score':<10}")
print("-" * 50)

# Print each student in descending order
for rank, (student, score) in enumerate(student_score_pairs, 1):
    print(f"{rank:<6} {student:<15} {score:<10}")

print("-" * 50)
print(f"\nTotal Class Score: {total_class_score}")
print(f"Class Average: {total_class_score / len(all_scores):.2f}")
print(f"Class Highest: {max(all_scores)}")
print(f"Class Lowest: {min(all_scores)}")
print(f"Number of Students: {len(all_students)}")
```
