```python
#Assignment Operator 
F_Score = float(input("Please enter your First Semester Score : "))
S_Score = float(input("please enter your secound Semester Score : "))
```

```python
#Arithmetic Operator 
avg = (F_Score+S_Score)/2
print("The Average Score is :", avg)
```

```python
#Comparison OPerator 
if F_Score > S_Score:
    print("First score higher then Secound score")
else:
    print("Secound score higher then Secound score")
```

```python
# Using logical AND operator
if F_Score >= 90 and S_Score >= 90:
    print("Both Semester, First Semester and Secound Semester have excellent scores")
else: 
    print("Both Semester, First Semester and Secound Semester have not Good scores")
```
