## Question 1: Sort the dataset

```python
import random
import time

# Generate dataset (given code)
random.seed(10)
data = random.sample(range(1, 20000), 10000)

# Sort the dataset
print(data)
data.sort()
print(f"Printing sorted Data : {data}")
```

## Question 2: Implement Sequential Search
#### Complexity:
#### Best Case: O(1)
#### Worst Case: O(n)

```python
def sequential_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

data = [10, 20, 30, 40, 50]
result1 = sequential_search(data, 30)
print("index:", result1)
```

## Question 3: Implement Recursive Binary Search
#### Complexity:
#### Best Case: O(1)
#### Worst Case: O(log n)

```python
def binary_search_recursive(arr, target, left, right):
    if left > right:
        return -1
    
    mid = (left + right) // 2
    
    if arr[mid] == target:
        return mid
    elif arr[mid] > target:
        return binary_search_recursive(arr, target, left, mid - 1)
    else:
        return binary_search_recursive(arr, target, mid + 1, right)
        
result = binary_search_recursive(data, 50, 0, len(data) - 1)
```

## Question 4: Select random target from dataset

```python
target = random.choice(data)
print("Target:", target)
```

## Question 5: Measure execution time using time.monotonic()
#### Here is 2 part : 
(a): Sequential Search timing
(b): Binary Search timing

```python
# (a): Sequential Search timing
start = time.monotonic()
seq_result = sequential_search(data, target)
end = time.monotonic()

seq_time = end - start

print("Sequential Search:")
print("Index:", seq_result)
print("Time:", seq_time)

# (b): Binary Search timing
start = time.monotonic()
bin_result = binary_search_recursive(data, target, 0, len(data) - 1)
end = time.monotonic()

bin_time = end - start

print("Binary Search:")
print("Index:", bin_result)
print("Time:", bin_time)
```

## Question 6: Compare performance

```python
print("Performance Comparison..............!")

if seq_time > bin_time:
    print("Binary Search is faster.")
else:
    print("Sequential Search is faster.")
```
