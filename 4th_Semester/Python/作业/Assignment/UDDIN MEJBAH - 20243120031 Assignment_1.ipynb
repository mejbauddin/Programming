## 1: Get 3 inputs from user
Ask user for:
1. An integer (whole number like 5, -10, 100)
2. A float (decimal number like 3.14, 2.5, -0.5)
3. A string (text like "Hello", "123", "Python")

```python
# Get integer input
user_int = int(input("Enter an integer: "))
# Get float input
user_float = float(input("Enter a float: "))
# Get string input
user_string = input("Enter a string: ")
```

## Question 2: Print type() and id() of each variable
#### What is type()?
Tells you what kind of data it is (int, float, str)
#### What is id()?
Gives a unique number that identifies where Python stores this data in memory

```python
# integer Value
print(f"Integer value: {user_int}")
print(f"  Type: {type(user_int)}")
print(f"  ID: {id(user_int)}")

# float Value
print(f"\nFloat value: {user_float}")
print(f"  Type: {type(user_float)}")
print(f"  ID: {id(user_float)}")

# String Value
print(f"\nString value: '{user_string}'")
print(f"  Type: {type(user_string)}")
print(f"  ID: {id(user_string)}")
```

## Question 3: Convert integer to binary and display bit length
#### What is binary?
Computers understand only 0s and 1s
Binary is how numbers look to computer: 5 = 101, 10 = 1010
#### What is bin()?
Converts number to binary format
#### What is bit length?
How many binary digits (0/1) needed to represent the number
Example: 5 in binary is 101 → 3 bits

```python
binary_representation = bin(user_int)
bit_length = user_int.bit_length()
print(f"Integer: {user_int}")
print(f"Binary representation: {binary_representation}")
print(f"Bit length: {bit_length} bits")
```

## Question 4: Convert the float to an integer explicitly
#### What does this mean?
Change a decimal number to a whole number
Python just cuts off the decimal part (does NOT round)

```python
float_to_int = int(user_float)
print(f"Original float: {user_float}")
print(f"Converted to integer: {float_to_int}")
```

## Question 5: Determine if string can be converted to integer safely
#### What does this mean?
Check if text like "123" can become number 123
But "Hello" cannot become a number
#### How to check?
Try to convert, if it fails → catch the error

```python
try:
    string_to_int = int(user_string)
    print(f"String '{user_string}' CAN be converted to integer: {string_to_int}")
except ValueError:
    print(f"String '{user_string}' CANNOT be converted to integer safely")
    print(f"Reason: String contains non-numeric characters or invalid format")
```

## Question 6: Display Boolean evaluation of each input
#### What is Boolean?
True or False values
Python decides if something is True or False based on its value
#### Simple Rules:
False values: 0, 0.0, empty string "", None
True values: Everything else

```python
# Boolean evaluation of integer
bool_int = bool(user_int)
print(f"Boolean evaluation of integer {user_int}: {bool_int}")

# Boolean evaluation of float
bool_float = bool(user_float)
print(f"Boolean evaluation of float {user_float}: {bool_float}")

# Boolean evaluation of string
bool_string = bool(user_string)
print(f"Boolean evaluation of string '{user_string}': {bool_string}")
```
